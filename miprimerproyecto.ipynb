**FUNCIONES EN PYTHON**

```python
def suma(num1, num2):
  respuesta = float(num1) + float(num2)
  return respuesta
```

```python
def resta(num1, num2):
  respuesta = float(num1) - float(num2)
  return respuesta
```

```python
num1 = input("num 1: ")
num2 = input("num 2: ")

print("Respuesta: ", suma(num1, num2))
print("Respuesta: ", resta(num1, num2))
```

Hacer el siguiente ejercicio en python haciendo uso de funciones, que convierta grados celcius a Ferenheit

```python
def convertir(celcius):
  return float(celcius)*9/5 + 32
```

```python
c = input("grados Celcius: ")
print("grados Ferenheit:", convertir(c))
```

Hacer el siguiente ejercicio en python haciendo uso de funciones, que convierta grados Farenheit a Celcius

```python
def converFac(f):
  return (float(f)-32)*5/9
```

```python
f = input("F: ")
print("Grados Celcius: ", converFac(f))
```

Hacer un ejercicio en python haciendo uso de funciones de usuario, que convierta grados Celcius a Kelvin

```python
def converCak(c):
  return float(c) + 273.15
```

```python
c = input("C: ")
print("Grados Kelvin: ", converCak(c))
```

**FUNCIONES ANONIMAS**

Hcaer el siguiente ejercicio en python haciendo uso de funciones anonimas, que convierta grados celcius a Ferenheit

```python
convertir = lambda c: float(c)*9/5 + 32
```

```python
c = float(input("grados Celcius: "))
print("grados Ferenheit:", convertir(c))
```
