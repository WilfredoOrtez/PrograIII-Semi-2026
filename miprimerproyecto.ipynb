```python
def calculorSueldo(hrs, valor):
  salario_bruto = float(hrs) * float(valor) * 30
  isss = float(salario_bruto * 0.03)  # ISSS es el 3%

  if isss >= 30:
    isss = 30.0

  afp = float(salario_bruto * 0.0725)  # AFP es el 7.25%

  # Renta imponible = salario bruto menos ISSS y AFP
  renta_imponible = salario_bruto - (isss + afp)

  # El ISR se calcula sobre la renta imponible, no sobre el salario bruto
  isr = calcularIsr(renta_imponible)

  sueldo_final = renta_imponible - isr

  print(f"El salario bruto es: {salario_bruto}")
  print(f"El isss es: {isss}")
  print(f"El afp es: {afp}")
  print(f"La renta imponible es: {round(renta_imponible,2)}")
  print(f"El isr retenido es: {round(isr,2)}")
  print(f"El sueldo neto final es: {round(sueldo_final,2)}")


def calcularIsr(monto):
  monto = float(monto)
  # Calculamos el impuesto sobre renta (fórmula oficial: excedente + cuota fija)

  # Tramo 1
  if monto >= 0.01 and monto <= 472.00:
    return 0
  # Tramo 2
  elif monto >= 472.01 and monto <= 895.24:
    return (monto - 472.00) * 0.10 + 17.67
  #Tramo 3
  elif monto >= 895.25 and monto <= 2038.10:
    return (monto - 895.24) * 0.20 + 60.00
  else:
    # Tramo 4
    return (monto - 2038.10) * 0.30 + 288.57
```

```python
"""
Sistema Estadístico Académico
Calcula: Media, Mediana, Moda, Varianza, Desviación Estándar,
Desviación Típica, Rango y Frecuencia
"""

from functools import reduce
import json

# ==================== CLASE ESTADÍSTICA ====================

class SistemaEstadistico:
    """Calcula estadísticas básicas de una serie de datos"""

    def __init__(self, serie):
        # Convertimos todo a float
        self.serie = [float(x) for x in serie]

    def media(self):
        """Media aritmética: suma de todos los valores / cantidad de valores"""
        suma = reduce(lambda acum, num: acum + num, self.serie, 0)
        return suma / len(self.serie)

    def mediana(self):
        """Mediana: valor central cuando los datos están ordenados"""
        datos_ordenados = sorted(self.serie)
        n = len(datos_ordenados)
        mitad = n // 2
        if n % 2 == 0:
            return (datos_ordenados[mitad - 1] + datos_ordenados[mitad]) / 2
        else:
            return datos_ordenados[mitad]

    def frecuencia(self):
        """Frecuencia: cuántas veces se repite cada valor"""
        frec = {}
        for valor in self.serie:
            frec[valor] = frec.get(valor, 0) + 1
        return frec

    def moda(self):
        """Moda: valor que más se repite"""
        frec = self.frecuencia()
        frecuencia_maxima = max(frec.values())
        if frecuencia_maxima == 1:
            return None  # No hay moda
        modas = [valor for valor, cant in frec.items() if cant == frecuencia_maxima]
        return modas

    def varianza(self):
        """Varianza: promedio de las diferencias al cuadrado respecto a la media"""
        media = self.media()
        suma = reduce(lambda acum, num: acum + (num - media) ** 2, self.serie, 0)
        return suma / len(self.serie)

    def desviacion_estandar(self):
        """Desviación estándar: raíz cuadrada de la varianza"""
        return self.varianza() ** 0.5

    def desviacion_tipica(self):
        """Desviación típica: sinónimo de desviación estándar"""
        return self.desviacion_estandar()

    def rango(self):
        """Rango: diferencia entre el valor máximo y mínimo"""
        return max(self.serie) - min(self.serie)

    def resumen(self):
        """Muestra todas las estadísticas calculadas"""
        print("\n" + "="*60)
        print(" RESUMEN ESTADÍSTICO")
        print("="*60)
        print(f"Serie de datos: {self.serie}")
        print(f"Media aritmética: {round(self.media(), 2)}")
        print(f"Mediana: {round(self.mediana(), 2)}")
        moda = self.moda()
        print(f"Moda: {moda if moda else 'No hay moda (todos los valores son únicos)'}")
        print(f"Varianza: {round(self.varianza(), 2)}")
        print(f"Desviación estándar: {round(self.desviacion_estandar(), 2)}")
        print(f"Desviación típica: {round(self.desviacion_tipica(), 2)}")
        print(f"Rango: {round(self.rango(), 2)}")
        print("Frecuencia:", self.frecuencia())


# ==================== CLASE ESTUDIANTE ====================

class Estudiante:
    """Representa a un estudiante con sus datos académicos"""

    def __init__(self, nombre, materias, notas_lab1, notas_lab2, nota_parcial,
                 edad, tiempo_traslado, horas_telefono):
        self.nombre = nombre
        self.materias = materias  # Lista de materias
        self.notas_lab1 = notas_lab1  # Lista de notas del laboratorio 1
        self.notas_lab2 = notas_lab2  # Lista de notas del laboratorio 2
        self.nota_parcial = nota_parcial  # Lista de notas del parcial
        self.edad = edad
        self.tiempo_traslado = tiempo_traslado  # minutos
        self.horas_telefono = horas_telefono  # horas diarias

    def calcular_nota_final(self, indice_materia):
        """Calcula la nota final de una materia específica (30% lab1 + 30% lab2 + 40% parcial)"""
        lab1 = self.notas_lab1[indice_materia] * 0.30
        lab2 = self.notas_lab2[indice_materia] * 0.30
        parcial = self.nota_parcial[indice_materia] * 0.40
        return lab1 + lab2 + parcial

    def obtener_notas_finales(self):
        """Obtiene la lista de todas las notas finales del estudiante"""
        return [self.calcular_nota_final(i) for i in range(len(self.materias))]

    def promedio_notas(self):
        """Calcula el promedio de las notas finales"""
        notas = self.obtener_notas_finales()
        return reduce(lambda acum, n: acum + n, notas, 0) / len(notas)

    def aprobado(self):
        """Determina si el estudiante está aprobado (promedio >= 6)"""
        return self.promedio_notas() >= 6

    def to_dict(self):
        """Convierte los datos del estudiante a diccionario"""
        notas_finales = self.obtener_notas_finales()
        return {
            "nombre": self.nombre,
            "materias": self.materias,
            "notas_lab1": self.notas_lab1,
            "notas_lab2": self.notas_lab2,
            "nota_parcial": self.nota_parcial,
            "notas_finales": [round(n, 2) for n in notas_finales],
            "promedio": round(self.promedio_notas(), 2),
            "estado": "Aprobado" if self.aprobado() else "Reprobado",
            "edad": self.edad,
            "tiempo_traslado_min": self.tiempo_traslado,
            "horas_telefono": self.horas_telefono
        }


# ==================== SISTEMA ACADÉMICO ====================

class SistemaAcademico:
    """Administra una lista de estudiantes y calcula estadísticas del grupo"""

    def __init__(self):
        self.estudiantes = []

    def agregar_estudiante(self, estudiante):
        """Agrega un nuevo estudiante al sistema"""
        self.estudiantes.append(estudiante)
        print(f" Estudiante '{estudiante.nombre}' agregado exitosamente!")

    def eliminar_estudiante(self, nombre):
        """Elimina un estudiante por su nombre"""
        for i, est in enumerate(self.estudiantes):
            if est.nombre.lower() == nombre.lower():
                self.estudiantes.pop(i)
                print(f" Estudiante '{nombre}' eliminado exitosamente!")
                return True
        print(f" No se encontró al estudiante '{nombre}'")
        return False

    def listar_estudiantes(self):
        """Muestra el listado completo de estudiantes"""
        if not self.estudiantes:
            print(" No hay estudiantes en el sistema")
            return

        print("\n" + "="*80)
        print(" LISTADO DE ESTUDIANTES")
        print("="*80)
        for i, est in enumerate(self.estudiantes, 1):
            datos = est.to_dict()
            print(f"\n{i}. {datos['nombre']}")
            print(f"    Materias: {', '.join(datos['materias'])}")
            print(f"    Notas finales: {datos['notas_finales']}")
            print(f"    Promedio: {datos['promedio']} - {datos['estado']}")
            print(f"    Edad: {datos['edad']} años")
            print(f"    Tiempo traslado: {datos['tiempo_traslado_min']} min")
            print(f"    Horas teléfono: {datos['horas_telefono']}h")
            print("-"*40)

    def obtener_todas_las_notas(self):
        """Obtiene todas las notas finales de todos los estudiantes"""
        todas_notas = []
        for est in self.estudiantes:
            todas_notas.extend(est.obtener_notas_finales())
        return todas_notas

    def obtener_todas_edades(self):
        """Obtiene todas las edades de los estudiantes"""
        return [est.edad for est in self.estudiantes]

    def obtener_todos_traslados(self):
        """Obtiene todos los tiempos de traslado"""
        return [est.tiempo_traslado for est in self.estudiantes]

    def obtener_todas_horas_telefono(self):
        """Obtiene todas las horas de uso del teléfono"""
        return [est.horas_telefono for est in self.estudiantes]

    def obtener_todos_promedios(self):
        """Obtiene todos los promedios de los estudiantes"""
        return [est.promedio_notas() for est in self.estudiantes]

    def estudiantes_por_materia(self, materia):
        """Filtra estudiantes por materia"""
        return [est for est in self.estudiantes
                if any(materia.lower() in m.lower() for m in est.materias)]

    def aprobados(self):
        """Lista de estudiantes aprobados"""
        return [est for est in self.estudiantes if est.aprobado()]

    def reprobados(self):
        """Lista de estudiantes reprobados"""
        return [est for est in self.estudiantes if not est.aprobado()]

    def a_json(self):
        """Convierte todos los datos a formato JSON"""
        return json.dumps([est.to_dict() for est in self.estudiantes],
                         indent=2, ensure_ascii=False)


# ==================== DATOS INICIALES ====================

def cargar_datos_iniciales():
    """Carga los 20 estudiantes con las materias y estructura de notas correcta"""

    # Lista de materias correctas
    materias = [
        "Bases de Datos II",
        "Inglés Básico",
        "Programación Computacional III",
        "Diseño de Circuitos Digitales",
        "Estadística"
    ]

    # Datos de los estudiantes: nombre, [notas lab1], [notas lab2], [notas parcial], edad, traslado, horas_tel
    datos_estudiantes = [
        # Nombre, [Lab1], [Lab2], [Parcial], Edad, Traslado, Horas Tel
        ("Edgar Ernesto Ramirez Barrera", [8, 9, 7, 8, 9], [7, 8, 9, 8, 7], [9, 8, 8, 9, 8], 21, 2, 7),
        ("Yeimy Julissa Parada Chávez", [9, 8, 9, 9, 8], [9, 9, 8, 9, 9], [10, 9, 9, 10, 9], 20, 45, 3),
        ("Josué Stiven Funes Benavides", [8, 9, 8, 8, 9], [9, 8, 9, 9, 8], [9, 9, 9, 8, 9], 19, 30, 5),
        ("Wilfredo Josué Ortez Gaitan", [7, 8, 9, 8, 7], [8, 7, 8, 9, 8], [8, 9, 8, 7, 8], 20, 30, 5),
        ("Francisco Saravia", [8, 7, 8, 7, 8], [7, 8, 7, 8, 7], [8, 7, 8, 7, 8], 20, 35, 5),
        ("Jimmy Roberto Martines Gomez", [6, 7, 6, 7, 6], [7, 6, 7, 6, 7], [6, 7, 6, 7, 6], 19, 20, 3),
    ]

    sistema = SistemaAcademico()
    for nombre, lab1, lab2, parcial, edad, traslado, horas_tel in datos_estudiantes:
        estudiante = Estudiante(
            nombre,
            materias,
            lab1,
            lab2,
            parcial,
            edad,
            traslado,
            horas_tel
        )
        sistema.agregar_estudiante(estudiante)

    return sistema


# ==================== FUNCIONES DE MENÚ ====================

def mostrar_estadisticas_variable(titulo, datos):
    """Muestra el resumen estadístico de una variable"""
    print(f"\n {titulo}")
    print("-"*60)
    if not datos:
        print("No hay datos disponibles")
        return
    estadisticas = SistemaEstadistico(datos)
    estadisticas.resumen()


def mostrar_menu_principal():
    """Muestra el menú principal"""
    print("\n" + "="*60)
    print(" SISTEMA ESTADÍSTICO ACADÉMICO - UGB")
    print("="*60)
    print("1.  Listar todos los estudiantes")
    print("2.  Estadísticas de todas las notas")
    print("3.  Estadísticas de edades")
    print("4.  Estadísticas de tiempo de traslado")
    print("5.  Estadísticas de horas de teléfono")
    print("6.  Estadísticas de promedios")
    print("7.  Ver aprobados y reprobados")
    print("8.  Agregar nuevo estudiante")
    print("9.  Eliminar estudiante")
    print("10.  Buscar estudiantes por materia")
    print("11.  Ver datos en JSON")
    print("0.  Salir")
    print("="*60)


def agregar_nuevo_estudiante(sistema):
    """Permite agregar un nuevo estudiante al sistema"""
    print("\n AGREGAR NUEVO ESTUDIANTE")
    print("-"*40)

    nombre = input("Nombre completo: ").strip()

    # Materias disponibles
    materias_disponibles = [
        "Bases de Datos II",
        "Inglés Básico",
        "Programación Computacional III",
        "Diseño de Circuitos Digitales",
        "Estadística"
    ]

    print("\nMaterias disponibles:")
    for i, m in enumerate(materias_disponibles, 1):
        print(f"{i}. {m}")

    # Seleccionar materias
    seleccion = input("\nIngrese números de materias (separados por coma, ej: 1,3,5): ").strip()
    indices = [int(x.strip()) - 1 for x in seleccion.split(",") if x.strip().isdigit()]
    materias_seleccionadas = [materias_disponibles[i] for i in indices if 0 <= i < len(materias_disponibles)]

    if not materias_seleccionadas:
        print(" No seleccionó materias válidas")
        return

    print(f"\n Materias seleccionadas: {', '.join(materias_seleccionadas)}")

    # Ingresar notas (3 notas por materia)
    notas_lab1 = []
    notas_lab2 = []
    notas_parcial = []

    for materia in materias_seleccionadas:
        print(f"\n--- {materia} ---")
        lab1 = float(input("Nota Laboratorio 1 (0-10): "))
        lab2 = float(input("Nota Laboratorio 2 (0-10): "))
        parcial = float(input("Nota Parcial (0-10): "))
        notas_lab1.append(lab1)
        notas_lab2.append(lab2)
        notas_parcial.append(parcial)

    edad = int(input("\n Edad: "))
    traslado = float(input(" Tiempo de traslado a la UGB (minutos): "))
    horas_tel = float(input(" Horas de uso del teléfono diario: "))

    # Crear y agregar estudiante
    estudiante = Estudiante(nombre, materias_seleccionadas, notas_lab1, notas_lab2,
                           notas_parcial, edad, traslado, horas_tel)
    sistema.agregar_estudiante(estudiante)


def mostrar_aprobados_reprobados(sistema):
    """Muestra la lista de aprobados y reprobados"""
    aprobados = sistema.aprobados()
    reprobados = sistema.reprobados()

    print("\n ESTADÍSTICAS DE APROBACIÓN")
    print("="*60)
    print(f" Aprobados: {len(aprobados)} estudiantes")
    for est in aprobados:
        print(f"   - {est.nombre}: {est.promedio_notas():.2f}")

    print(f"\n Reprobados: {len(reprobados)} estudiantes")
    for est in reprobados:
        print(f"   - {est.nombre}: {est.promedio_notas():.2f}")


def buscar_por_materia(sistema):
    """Busca estudiantes por materia"""
    print("\n BUSCAR ESTUDIANTES POR MATERIA")
    materia = input("Ingrese el nombre de la materia: ").strip()

    resultados = sistema.estudiantes_por_materia(materia)

    if not resultados:
        print(f"No se encontraron estudiantes en '{materia}'")
        return

    print(f"\n Estudiantes en '{materia}': {len(resultados)}")
    for est in resultados:
        datos = est.to_dict()
        # Encontrar la nota de esa materia
        for i, m in enumerate(est.materias):
            if materia.lower() in m.lower():
                nota = datos['notas_finales'][i]
                print(f"   - {est.nombre}: {nota:.2f}")


# ==================== PROGRAMA PRINCIPAL ====================

def main():
    """Función principal del programa"""
    print("\n🎓 INICIANDO SISTEMA ACADÉMICO UGB...")
    sistema = cargar_datos_iniciales()
    print(f" {len(sistema.estudiantes)} estudiantes cargados exitosamente!")

    while True:
        mostrar_menu_principal()
        opcion = input("\n🔹 Elija una opción: ").strip()

        if opcion == "0":
            print(" ¡Hasta luego! Gracias por usar el sistema.")
            break

        elif opcion == "1":
            sistema.listar_estudiantes()

        elif opcion == "2":
            notas = sistema.obtener_todas_las_notas()
            mostrar_estadisticas_variable("ESTADÍSTICAS DE TODAS LAS NOTAS", notas)

        elif opcion == "3":
            edades = sistema.obtener_todas_edades()
            mostrar_estadisticas_variable("ESTADÍSTICAS DE EDADES", edades)

        elif opcion == "4":
            traslados = sistema.obtener_todos_traslados()
            mostrar_estadisticas_variable("ESTADÍSTICAS DE TIEMPO DE TRASLADO", traslados)

        elif opcion == "5":
            horas = sistema.obtener_todas_horas_telefono()
            mostrar_estadisticas_variable("ESTADÍSTICAS DE HORAS DE TELÉFONO", horas)

        elif opcion == "6":
            promedios = sistema.obtener_todos_promedios()
            mostrar_estadisticas_variable("ESTADÍSTICAS DE PROMEDIOS", promedios)

        elif opcion == "7":
            mostrar_aprobados_reprobados(sistema)

        elif opcion == "8":
            agregar_nuevo_estudiante(sistema)

        elif opcion == "9":
            print("\n ELIMINAR ESTUDIANTE")
            nombre = input("Nombre del estudiante a eliminar: ").strip()
            confirmar = input(f"¿Está seguro de eliminar a '{nombre}'? (s/n): ").lower()
            if confirmar == 's':
                sistema.eliminar_estudiante(nombre)

        elif opcion == "10":
            buscar_por_materia(sistema)

        elif opcion == "11":
            print("\n DATOS EN FORMATO JSON")
            print("="*60)
            print(sistema.a_json())

        else:
            print(" Opción no válida. Intente nuevamente.")

        input("\n Presione Enter para continuar...")


if __name__ == "__main__":
    main()
```
