```python
def calculorSueldo(hrs, valor):
  salario_bruto = float(hrs) * float(valor) * 30
  isss = float(salario_bruto * 0.03)  # ISSS es el 3%

  if isss >= 30:
    isss = 30.0

  afp = float(salario_bruto * 0.0725)  # AFP es el 7.25%

  # Renta imponible = salario bruto menos ISSS y AFP
  renta_imponible = salario_bruto - (isss + afp)

  # El ISR se calcula sobre la renta imponible, no sobre el salario bruto
  isr = calcularIsr(renta_imponible)

  sueldo_final = renta_imponible - isr

  print(f"El salario bruto es: {salario_bruto}")
  print(f"El isss es: {isss}")
  print(f"El afp es: {afp}")
  print(f"La renta imponible es: {round(renta_imponible,2)}")
  print(f"El isr retenido es: {round(isr,2)}")
  print(f"El sueldo neto final es: {round(sueldo_final,2)}")


def calcularIsr(monto):
  monto = float(monto)
  # Calculamos el impuesto sobre renta (fórmula oficial: excedente + cuota fija)

  # Tramo 1
  if monto >= 0.01 and monto <= 472.00:
    return 0
  # Tramo 2
  elif monto >= 472.01 and monto <= 895.24:
    return (monto - 472.00) * 0.10 + 17.67
  #Tramo 3
  elif monto >= 895.25 and monto <= 2038.10:
    return (monto - 895.24) * 0.20 + 60.00
  else:
    # Tramo 4
    return (monto - 2038.10) * 0.30 + 288.57
```

```python
"""
Sistema Estadístico Académico
Calcula: Media, Mediana, Moda, Varianza, Desviación Estándar,
Desviación Típica, Rango y Frecuencia
"""

from functools import reduce
import json

# ==================== CLASE ESTADÍSTICA ====================

class SistemaEstadistico:
    """Calcula estadísticas básicas de una serie de datos"""

    def __init__(self, serie):
        # Convertimos todo a float
        self.serie = [float(x) for x in serie]

    def media(self):
        """Media aritmética: suma de todos los valores / cantidad de valores"""
        suma = reduce(lambda acum, num: acum + num, self.serie, 0)
        return suma / len(self.serie)

    def mediana(self):
        """Mediana: valor central cuando los datos están ordenados"""
        datos_ordenados = sorted(self.serie)
        n = len(datos_ordenados)
        mitad = n // 2
        if n % 2 == 0:
            return (datos_ordenados[mitad - 1] + datos_ordenados[mitad]) / 2
        else:
            return datos_ordenados[mitad]

    def frecuencia(self):
        """Frecuencia: cuántas veces se repite cada valor"""
        frec = {}
        for valor in self.serie:
            frec[valor] = frec.get(valor, 0) + 1
        return frec

    def moda(self):
        """Moda: valor que más se repite"""
        frec = self.frecuencia()
        frecuencia_maxima = max(frec.values())
        if frecuencia_maxima == 1:
            return None  # No hay moda
        modas = [valor for valor, cant in frec.items() if cant == frecuencia_maxima]
        return modas

    def varianza(self):
        """Varianza: promedio de las diferencias al cuadrado respecto a la media"""
        media = self.media()
        suma = reduce(lambda acum, num: acum + (num - media) ** 2, self.serie, 0)
        return suma / len(self.serie)

    def desviacion_estandar(self):
        """Desviación estándar: raíz cuadrada de la varianza"""
        return self.varianza() ** 0.5

    def desviacion_tipica(self):
        """Desviación típica: sinónimo de desviación estándar"""
        return self.desviacion_estandar()

    def rango(self):
        """Rango: diferencia entre el valor máximo y mínimo"""
        return max(self.serie) - min(self.serie)

    def resumen(self):
        """Muestra todas las estadísticas calculadas"""
        print("\n" + "="*60)
        print(" RESUMEN ESTADÍSTICO")
        print("="*60)
        print(f"Serie de datos: {self.serie}")
        print(f"Media aritmética: {round(self.media(), 2)}")
        print(f"Mediana: {round(self.mediana(), 2)}")
        moda = self.moda()
        print(f"Moda: {moda if moda else 'No hay moda (todos los valores son únicos)'}")
        print(f"Varianza: {round(self.varianza(), 2)}")
        print(f"Desviación estándar: {round(self.desviacion_estandar(), 2)}")
        print(f"Desviación típica: {round(self.desviacion_tipica(), 2)}")
        print(f"Rango: {round(self.rango(), 2)}")
        print("Frecuencia:", self.frecuencia())


# ==================== CLASE ESTUDIANTE ====================

class Estudiante:
    """Representa a un estudiante con sus datos académicos"""

    def __init__(self, nombre, materias, notas_lab1, notas_lab2, nota_parcial,
                 edad, tiempo_traslado, horas_telefono):
        self.nombre = nombre
        self.materias = materias  # Lista de materias
        self.notas_lab1 = notas_lab1  # Lista de notas del laboratorio 1
        self.notas_lab2 = notas_lab2  # Lista de notas del laboratorio 2
        self.nota_parcial = nota_parcial  # Lista de notas del parcial
        self.edad = edad
        self.tiempo_traslado = tiempo_traslado  # minutos
        self.horas_telefono = horas_telefono  # horas diarias

    def calcular_nota_final(self, indice_materia):
        """Calcula la nota final de una materia específica (30% lab1 + 30% lab2 + 40% parcial)"""
        lab1 = self.notas_lab1[indice_materia] * 0.30
        lab2 = self.notas_lab2[indice_materia] * 0.30
        parcial = self.nota_parcial[indice_materia] * 0.40
        return lab1 + lab2 + parcial

    def obtener_notas_finales(self):
        """Obtiene la lista de todas las notas finales del estudiante"""
        return [self.calcular_nota_final(i) for i in range(len(self.materias))]

    def promedio_notas(self):
        """Calcula el promedio de las notas finales"""
        notas = self.obtener_notas_finales()
        return reduce(lambda acum, n: acum + n, notas, 0) / len(notas)

    def aprobado(self):
        """Determina si el estudiante está aprobado (promedio >= 6)"""
        return self.promedio_notas() >= 6

    def to_dict(self):
        """Convierte los datos del estudiante a diccionario"""
        notas_finales = self.obtener_notas_finales()
        return {
            "nombre": self.nombre,
            "materias": self.materias,
            "notas_lab1": self.notas_lab1,
            "notas_lab2": self.notas_lab2,
            "nota_parcial": self.nota_parcial,
            "notas_finales": [round(n, 2) for n in notas_finales],
            "promedio": round(self.promedio_notas(), 2),
            "estado": "Aprobado" if self.aprobado() else "Reprobado",
            "edad": self.edad,
            "tiempo_traslado_min": self.tiempo_traslado,
            "horas_telefono": self.horas_telefono
        }


# ==================== SISTEMA ACADÉMICO ====================

class SistemaAcademico:
    """Administra una lista de estudiantes y calcula estadísticas del grupo"""

    def __init__(self):
        self.estudiantes = []

    def agregar_estudiante(self, estudiante):
        """Agrega un nuevo estudiante al sistema"""
        self.estudiantes.append(estudiante)
        print(f" Estudiante '{estudiante.nombre}' agregado exitosamente!")

    def eliminar_estudiante(self, nombre):
        """Elimina un estudiante por su nombre"""
        for i, est in enumerate(self.estudiantes):
            if est.nombre.lower() == nombre.lower():
                self.estudiantes.pop(i)
                print(f" Estudiante '{nombre}' eliminado exitosamente!")
                return True
        print(f" No se encontró al estudiante '{nombre}'")
        return False

    def listar_estudiantes(self):
        """Muestra el listado completo de estudiantes"""
        if not self.estudiantes:
            print(" No hay estudiantes en el sistema")
            return

        print("\n" + "="*80)
        print(" LISTADO DE ESTUDIANTES")
        print("="*80)
        for i, est in enumerate(self.estudiantes, 1):
            datos = est.to_dict()
            print(f"\n{i}. {datos['nombre']}")
            print(f"    Materias: {', '.join(datos['materias'])}")
            print(f"    Notas finales: {datos['notas_finales']}")
            print(f"    Promedio: {datos['promedio']} - {datos['estado']}")
            print(f"    Edad: {datos['edad']} años")
            print(f"    Tiempo traslado: {datos['tiempo_traslado_min']} min")
            print(f"    Horas teléfono: {datos['horas_telefono']}h")
            print("-"*40)

    def obtener_todas_las_notas(self):
        """Obtiene todas las notas finales de todos los estudiantes"""
        todas_notas = []
        for est in self.estudiantes:
            todas_notas.extend(est.obtener_notas_finales())
        return todas_notas

    def obtener_todas_edades(self):
        """Obtiene todas las edades de los estudiantes"""
        return [est.edad for est in self.estudiantes]

    def obtener_todos_traslados(self):
        """Obtiene todos los tiempos de traslado"""
        return [est.tiempo_traslado for est in self.estudiantes]

    def obtener_todas_horas_telefono(self):
        """Obtiene todas las horas de uso del teléfono"""
        return [est.horas_telefono for est in self.estudiantes]

    def obtener_todos_promedios(self):
        """Obtiene todos los promedios de los estudiantes"""
        return [est.promedio_notas() for est in self.estudiantes]

    def estudiantes_por_materia(self, materia):
        """Filtra estudiantes por materia"""
        return [est for est in self.estudiantes
                if any(materia.lower() in m.lower() for m in est.materias)]

    def aprobados(self):
        """Lista de estudiantes aprobados"""
        return [est for est in self.estudiantes if est.aprobado()]

    def reprobados(self):
        """Lista de estudiantes reprobados"""
        return [est for est in self.estudiantes if not est.aprobado()]

    def a_json(self):
        """Convierte todos los datos a formato JSON"""
        return json.dumps([est.to_dict() for est in self.estudiantes],
                         indent=2, ensure_ascii=False)


# ==================== DATOS INICIALES ====================

def cargar_datos_iniciales():
    """Carga los 20 estudiantes con las materias y estructura de notas correcta"""

    # Lista de materias correctas
    materias = [
        "Bases de Datos II",
        "Inglés Básico",
        "Programación Computacional III",
        "Diseño de Circuitos Digitales",
        "Estadística"
    ]

    # Datos de los estudiantes: nombre, [notas lab1], [notas lab2], [notas parcial], edad, traslado, horas_tel
    datos_estudiantes = [
        # Nombre, [Lab1], [Lab2], [Parcial], Edad, Traslado, Horas Tel
        ("Edgar Ernesto Ramirez Barrera", [8, 9, 7, 8, 9], [7, 8, 9, 8, 7], [9, 8, 8, 9, 8], 21, 2, 7),
        ("Yeimy Julissa Parada Chávez", [9, 8, 9, 9, 8], [9, 9, 8, 9, 9], [10, 9, 9, 10, 9], 20, 45, 3),
        ("Josué Stiven Funes Benavides", [8, 9, 8, 8, 9], [9, 8, 9, 9, 8], [9, 9, 9, 8, 9], 19, 30, 5),
        ("Wilfredo Josué Ortez Gaitan", [7, 8, 9, 8, 7], [8, 7, 8, 9, 8], [8, 9, 8, 7, 8], 20, 30, 5),
        ("Francisco Saravia", [8, 7, 8, 7, 8], [7, 8, 7, 8, 7], [8, 7, 8, 7, 8], 20, 35, 5),
        ("Jimmy Roberto Martines Gomez", [6, 7, 6, 7, 6], [7, 6, 7, 6, 7], [6, 7, 6, 7, 6], 19, 20, 3),
    ]

    sistema = SistemaAcademico()
    for nombre, lab1, lab2, parcial, edad, traslado, horas_tel in datos_estudiantes:
        estudiante = Estudiante(
            nombre,
            materias,
            lab1,
            lab2,
            parcial,
            edad,
            traslado,
            horas_tel
        )
        sistema.agregar_estudiante(estudiante)

    return sistema


# ==================== FUNCIONES DE MENÚ ====================

def mostrar_estadisticas_variable(titulo, datos):
    """Muestra el resumen estadístico de una variable"""
    print(f"\n {titulo}")
    print("-"*60)
    if not datos:
        print("No hay datos disponibles")
        return
    estadisticas = SistemaEstadistico(datos)
    estadisticas.resumen()


def mostrar_menu_principal():
    """Muestra el menú principal"""
    print("\n" + "="*60)
    print(" SISTEMA ESTADÍSTICO ACADÉMICO - UGB")
    print("="*60)
    print("1.  Listar todos los estudiantes")
    print("2.  Estadísticas de todas las notas")
    print("3.  Estadísticas de edades")
    print("4.  Estadísticas de tiempo de traslado")
    print("5.  Estadísticas de horas de teléfono")
    print("6.  Estadísticas de promedios")
    print("7.  Ver aprobados y reprobados")
    print("8.  Agregar nuevo estudiante")
    print("9.  Eliminar estudiante")
    print("10.  Buscar estudiantes por materia")
    print("11.  Ver datos en JSON")
    print("0.  Salir")
    print("="*60)


def agregar_nuevo_estudiante(sistema):
    """Permite agregar un nuevo estudiante al sistema"""
    print("\n AGREGAR NUEVO ESTUDIANTE")
    print("-"*40)

    nombre = input("Nombre completo: ").strip()

    # Materias disponibles
    materias_disponibles = [
        "Bases de Datos II",
        "Inglés Básico",
        "Programación Computacional III",
        "Diseño de Circuitos Digitales",
        "Estadística"
    ]

    print("\nMaterias disponibles:")
    for i, m in enumerate(materias_disponibles, 1):
        print(f"{i}. {m}")

    # Seleccionar materias
    seleccion = input("\nIngrese números de materias (separados por coma, ej: 1,3,5): ").strip()
    indices = [int(x.strip()) - 1 for x in seleccion.split(",") if x.strip().isdigit()]
    materias_seleccionadas = [materias_disponibles[i] for i in indices if 0 <= i < len(materias_disponibles)]

    if not materias_seleccionadas:
        print(" No seleccionó materias válidas")
        return

    print(f"\n Materias seleccionadas: {', '.join(materias_seleccionadas)}")

    # Ingresar notas (3 notas por materia)
    notas_lab1 = []
    notas_lab2 = []
    notas_parcial = []

    for materia in materias_seleccionadas:
        print(f"\n--- {materia} ---")
        lab1 = float(input("Nota Laboratorio 1 (0-10): "))
        lab2 = float(input("Nota Laboratorio 2 (0-10): "))
        parcial = float(input("Nota Parcial (0-10): "))
        notas_lab1.append(lab1)
        notas_lab2.append(lab2)
        notas_parcial.append(parcial)

    edad = int(input("\n Edad: "))
    traslado = float(input(" Tiempo de traslado a la UGB (minutos): "))
    horas_tel = float(input(" Horas de uso del teléfono diario: "))

    # Crear y agregar estudiante
    estudiante = Estudiante(nombre, materias_seleccionadas, notas_lab1, notas_lab2,
                           notas_parcial, edad, traslado, horas_tel)
    sistema.agregar_estudiante(estudiante)


def mostrar_aprobados_reprobados(sistema):
    """Muestra la lista de aprobados y reprobados"""
    aprobados = sistema.aprobados()
    reprobados = sistema.reprobados()

    print("\n ESTADÍSTICAS DE APROBACIÓN")
    print("="*60)
    print(f" Aprobados: {len(aprobados)} estudiantes")
    for est in aprobados:
        print(f"   - {est.nombre}: {est.promedio_notas():.2f}")

    print(f"\n Reprobados: {len(reprobados)} estudiantes")
    for est in reprobados:
        print(f"   - {est.nombre}: {est.promedio_notas():.2f}")


def buscar_por_materia(sistema):
    """Busca estudiantes por materia"""
    print("\n BUSCAR ESTUDIANTES POR MATERIA")
    materia = input("Ingrese el nombre de la materia: ").strip()

    resultados = sistema.estudiantes_por_materia(materia)

    if not resultados:
        print(f"No se encontraron estudiantes en '{materia}'")
        return

    print(f"\n Estudiantes en '{materia}': {len(resultados)}")
    for est in resultados:
        datos = est.to_dict()
        # Encontrar la nota de esa materia
        for i, m in enumerate(est.materias):
            if materia.lower() in m.lower():
                nota = datos['notas_finales'][i]
                print(f"   - {est.nombre}: {nota:.2f}")


# ==================== PROGRAMA PRINCIPAL ====================

def main():
    """Función principal del programa"""
    print("\n🎓 INICIANDO SISTEMA ACADÉMICO UGB...")
    sistema = cargar_datos_iniciales()
    print(f" {len(sistema.estudiantes)} estudiantes cargados exitosamente!")

    while True:
        mostrar_menu_principal()
        opcion = input("\n🔹 Elija una opción: ").strip()

        if opcion == "0":
            print(" ¡Hasta luego! Gracias por usar el sistema.")
            break

        elif opcion == "1":
            sistema.listar_estudiantes()

        elif opcion == "2":
            notas = sistema.obtener_todas_las_notas()
            mostrar_estadisticas_variable("ESTADÍSTICAS DE TODAS LAS NOTAS", notas)

        elif opcion == "3":
            edades = sistema.obtener_todas_edades()
            mostrar_estadisticas_variable("ESTADÍSTICAS DE EDADES", edades)

        elif opcion == "4":
            traslados = sistema.obtener_todos_traslados()
            mostrar_estadisticas_variable("ESTADÍSTICAS DE TIEMPO DE TRASLADO", traslados)

        elif opcion == "5":
            horas = sistema.obtener_todas_horas_telefono()
            mostrar_estadisticas_variable("ESTADÍSTICAS DE HORAS DE TELÉFONO", horas)

        elif opcion == "6":
            promedios = sistema.obtener_todos_promedios()
            mostrar_estadisticas_variable("ESTADÍSTICAS DE PROMEDIOS", promedios)

        elif opcion == "7":
            mostrar_aprobados_reprobados(sistema)

        elif opcion == "8":
            agregar_nuevo_estudiante(sistema)

        elif opcion == "9":
            print("\n ELIMINAR ESTUDIANTE")
            nombre = input("Nombre del estudiante a eliminar: ").strip()
            confirmar = input(f"¿Está seguro de eliminar a '{nombre}'? (s/n): ").lower()
            if confirmar == 's':
                sistema.eliminar_estudiante(nombre)

        elif opcion == "10":
            buscar_por_materia(sistema)

        elif opcion == "11":
            print("\n DATOS EN FORMATO JSON")
            print("="*60)
            print(sistema.a_json())

        else:
            print(" Opción no válida. Intente nuevamente.")

        input("\n Presione Enter para continuar...")


if __name__ == "__main__":
    main()
```

```python
import math

"""
SISTEMA ACADÉMICO Y ESTADÍSTICO INTERACTIVO - UGB (SIN JSON NI DICCIONARIOS)
"""

class Estudiante:
    """Clase que representa a un estudiante mediante atributos de objeto."""
    def __init__(self, id_est, nombre, materia, nota, edad, traslado, telefono):
        self.id = id_est
        self.nombre = nombre
        self.materia = materia
        self.nota = float(nota)
        self.edad = int(edad)
        self.traslado_min = int(traslado)
        self.uso_telefono_hrs = float(telefono)


class SistemaEstadistico:
    """Clase encargada del procesamiento estadístico usando listas y tuplas (sin diccionarios)."""
    def __init__(self, datos, nombre_variable="Variable"):
        self.datos = datos
        self.nombre_variable = nombre_variable
        self.n = len(datos)

    def calcular_media(self):
        if self.n == 0: return 0
        return sum(self.datos) / self.n

    def calcular_mediana(self):
        if self.n == 0: return 0
        datos_ordenados = sorted(self.datos)
        mitad = self.n // 2
        if self.n % 2 == 0:
            return (datos_ordenados[mitad - 1] + datos_ordenados[mitad]) / 2.0
        else:
            return datos_ordenados[mitad]

    def calcular_frecuencias(self):
        """Calcula frecuencias absolutas usando una lista de tuplas (elemento, conteo) en lugar de un diccionario."""
        frecuencias = []
        for dato in self.datos:
            encontrado = False
            for idx, (elem, count) in enumerate(frecuencias):
                if elem == dato:
                    frecuencias[idx] = (elem, count + 1)
                    encontrado = True
                    break
            if not encontrado:
                frecuencias.append((dato, 1))
        return frecuencias

    def calcular_moda(self):
        if self.n == 0: return []
        frecuencias = self.calcular_frecuencias()
        if not frecuencias:
            return []
        max_freq = max(count for elem, count in frecuencias)
        return [elem for elem, count in frecuencias if count == max_freq]

    def calcular_varianza(self, es_muestra=True):
        if self.n <= 1: return 0
        media = self.calcular_media()
        suma_cuadrados = sum((x - media) ** 2 for x in self.datos)
        divisor = (self.n - 1) if es_muestra else self.n
        return suma_cuadrados / divisor

    def calcular_desviacion_estandar(self, es_muestra=True):
        return math.sqrt(self.calcular_varianza(es_muestra))

    def calcular_rango(self):
        if self.n == 0: return 0
        return max(self.datos) - min(self.datos)

    def mostrar_reporte(self):
        print(f"\n" + "=" * 55)
        print(f" ANÁLISIS ESTADÍSTICO: {self.nombre_variable.upper()}")
        print("=" * 55)
        print(f"  • Total Registros (N)         : {self.n}")
        print(f"  • Media Aritmética            : {round(self.calcular_media(), 2)}")
        print(f"  • Mediana                     : {round(self.calcular_mediana(), 2)}")
        print(f"  • Moda                        : {self.calcular_moda()}")
        print(f"  • Varianza (Muestral)         : {round(self.calcular_varianza(), 2)}")
        print(f"  • Desviación Estándar / Típica: {round(self.calcular_desviacion_estandar(), 2)}")
        print(f"  • Rango                       : {round(self.calcular_rango(), 2)}")
        print(f"  • Frecuencias Absolutas       : {self.calcular_frecuencias()}")
        print("=" * 55)


class SistemaAcademico:
    """Clase para gestionar la colección de objetos Estudiante en memoria."""
    def __init__(self):
        self.estudiantes = []

    def precargar_estudiantes(self, lista_tuplas):
        for id_est, nombre, materia, nota, edad, traslado, telefono in lista_tuplas:
            nuevo = Estudiante(id_est, nombre, materia, nota, edad, traslado, telefono)
            self.estudiantes.append(nuevo)

    def agregar_estudiante(self, id_est, nombre, materia, nota, edad, traslado, telefono):
        nuevo = Estudiante(id_est, nombre, materia, nota, edad, traslado, telefono)
        self.estudiantes.append(nuevo)
        print(f"\n¡Estudiante '{nombre}' registrado exitosamente!")

    def extraer_variable(self, atributo):
        return [getattr(e, atributo) for e in self.estudiantes if hasattr(e, atributo)]

    def mostrar_dataset_consola(self):
        print(f"\n" + "=" * 80)
        print("                         LISTADO COMPLETO DE ESTUDIANTES")
        print("=" * 80)
        for e in self.estudiantes:
            print(f"ID: {e.id} | {e.nombre:<18} | Mat: {e.materia:<25} | Nota: {e.nota} | Edad: {e.edad} | Traslado: {e.traslado_min}m | Tel: {e.uso_telefono_hrs}h")
        print("=" * 80)


def ejecutar_menu_principal():
    # Dataset inicial usando tuplas en lugar de texto JSON
    tuplas_dataset = [
        ("001", "Samuel Portillo", "Estadística Computacional", 10.0, 20, 80, 6.0),
        ("002", "Carlos Turcios", "Admin. Bases de Datos II", 9.5, 20, 25, 5.0),
        ("003", "Gilda González", "Circuitos Digitales", 9.3, 19, 30, 5.9),
        ("004", "Saúl Jimenez", "Programación Comp. II", 10.0, 21, 60, 5.0),
        ("005", "Edgar Ramirez", "Inglés Básico", 10.0, 21, 120, 7.0),
        ("006", "Yeimy Parada", "Admin. Bases de Datos II", 9.5, 20, 45, 3.0),
        ("007", "Miguel Gómez", "Programación Comp. III", 9.5, 19, 15, 4.0),
        ("008", "Josué Funes", "Estadística Computacional", 9.5, 19, 30, 5.0),
        ("009", "Jefferson Castillo", "Circuitos Digitales", 9.5, 19, 15, 2.3),
        ("010", "Ramael Cordova", "Inglés Básico", 9.5, 22, 7, 13.5),
        ("011", "Carlos Parada", "Programación Comp. III", 8.5, 25, 10, 5.5),
        ("012", "Madison Valle", "Estadística Computacional", 10.0, 20, 30, 9.0),
        ("013", "Jennifer Renderos", "Admin. Bases de Datos II", 9.6, 21, 60, 3.0),
        ("014", "Diego Villafranco", "Circuitos Digitales", 9.0, 20, 30, 8.0),
        ("015", "Flor Campos", "Estadística Computacional", 9.0, 20, 6, 5.4),
        ("016", "Wilfredo Ortez", "Programación Comp. III", 9.5, 20, 30, 5.0),
        ("017", "Kimberly Castro", "Programación Comp. III", 10.0, 18, 20, 9.0)
    ]

    sistema = SistemaAcademico()
    sistema.precargar_estudiantes(tuplas_dataset)

    # Mapeo de opciones basado en listas/tuplas en lugar de un diccionario de control
    opciones_variables = [
        ("1", "nota", "Notas Académicas"),
        ("2", "edad", "Edades"),
        ("3", "traslado_min", "Tiempo de Traslado (Minutos)"),
        ("4", "uso_telefono_hrs", "Uso de Teléfono (Horas/Día)")
    ]

    while True:
        print("\n" + "=" * 55)
        print("        MENÚ SISTEMA ACADÉMICO Y ESTADÍSTICO UGB")
        print("=" * 55)
        print("  [1] Ver dataset completo (Formato de Objetos)")
        print("  [2] Calcular estadísticas de UNA variable específica")
        print("  [3] Calcular estadísticas de TODAS las variables")
        print("  [4] Registrar un nuevo estudiante")
        print("  [5] Salir del programa")
        print("=" * 55)

        opcion = input("Seleccione una opción (1-5): ").strip()

        if opcion == "1":
            sistema.mostrar_dataset_consola()

        elif opcion == "2":
            print("\n--- SELECCIONE LA VARIABLE A ANALIZAR ---")
            print("  1. Notas Académicas")
            print("  2. Edades")
            print("  3. Tiempo de Traslado a la UGB (Minutos)")
            print("  4. Uso de Teléfono (Horas/Día)")

            sub_opcion = input("Elija una variable (1-4): ").strip()
            encontrado = False
            for num_opt, clave, nombre in opciones_variables:
                if sub_opcion == num_opt:
                    datos = sistema.extraer_variable(clave)
                    analizador = SistemaEstadistico(datos, nombre_variable=nombre)
                    analizador.mostrar_reporte()
                    encontrado = True
                    break
            if not encontrado:
                print("\n[!] Opción de variable no válida.")

        elif opcion == "3":
            for _, clave, nombre in opciones_variables:
                datos = sistema.extraer_variable(clave)
                analizador = SistemaEstadistico(datos, nombre_variable=nombre)
                analizador.mostrar_reporte()

        elif opcion == "4":
            print("\n--- INGRESO DE NUEVO ESTUDIANTE ---")
            try:
                id_est = input("ID (ej. 018): ").strip()
                nombre = input("Nombre completo: ").strip()
                materia = input("Materia: ").strip()
                nota = float(input("Nota (0.0 - 10.0): "))
                edad = int(input("Edad: "))
                traslado = int(input("Tiempo de traslado en minutos: "))
                telefono = float(input("Horas de uso de teléfono al día: "))

                sistema.agregar_estudiante(id_est, nombre, materia, nota, edad, traslado, telefono)
            except ValueError:
                print("\n[!] Error: Asegúrese de ingresar valores numéricos válidos en nota, edad, traslado y teléfono.")

        elif opcion == "5":
            print("\nSaliendo del sistema... ¡Gracias!")
            break

        else:
            print("\n[!] Opción inválida. Intente de nuevo con un número del 1 al 5.")


if __name__ == "__main__":
    ejecutar_menu_principal()
```

```python
#Conversor

def conversor_area(valor, clave_origen, clave_destino):
    # Factores de conversión tomando como base 1 metro cuadrado (m2)
    # 1 manzana = 6,988.96 m2
    # 1 tarea salvadoreña (aprox. 1/16 de manzana) = ~436.81 m2
    factores_conversion = {
        "m2": 1.0,
        "km2": 1000000.0,
        "ha": 10000.0,
        "cm2": 0.0001,
        "v2": 0.698896,
        "mz": 6988.96,
        "tz": 436.81  # Tareas (medida utilizada en El Salvador)
    }

    # Lógica matemática central
    valor_en_m2 = valor * factores_conversion[clave_origen]
    resultado = valor_en_m2 / factores_conversion[clave_destino]

    return round(resultado, 4)

def mostrar_menu():
    print("\n" + "="*45)
    print("        CONVERSOR DE ÁREA (EL SALVADOR)")
    print("="*45)
    print("  [1] Metros cuadrados (m2)")
    print("  [2] Kilómetros cuadrados (km2)")
    print("  [3] Hectáreas (ha)")
    print("  [4] Centímetros cuadrados (cm2)")
    print("  [5] Varas cuadradas (v2)")
    print("  [6] Manzanas (mz)")
    print("  [7] Tareas (tz)")
    print("  [8] Salir del conversor")
    print("="*45)

# Diccionario para mapear la selección numérica del menú con las claves del cálculo
menu_unidades = {
    "1": ("m2", "Metros cuadrados"),
    "2": ("km2", "Kilómetros cuadrados"),
    "3": ("ha", "Hectáreas"),
    "4": ("cm2", "Centímetros cuadrados"),
    "5": ("v2", "Varas cuadradas"),
    "6": ("mz", "Manzanas"),
    "7": ("tz", "Tareas")
}

# --- Bloque de Ejecución (Bucle interactivo) ---
while True:
    mostrar_menu()

    opcion_origen = input("Selecciona el NÚMERO de la unidad de ORIGEN (1-8): ").strip()

    # Estructura de control para salir del bucle
    if opcion_origen == "8":
        print("\nCerrando el conversor de área. ¡Hasta luego!\n")
        break

    # Validamos que la opción exista
    if opcion_origen not in menu_unidades:
        print("\nError: Debes seleccionar un número válido del 1 al 8.")
        continue

    opcion_destino = input("Selecciona el NÚMERO de la unidad de DESTINO (1-7): ").strip()

    if opcion_destino not in menu_unidades:
        print("\nError: Debes seleccionar un número válido del 1 al 7 para el destino.")
        continue

    try:
        cantidad = float(input("\nIngresa la cantidad que deseas convertir: "))

        # Extraemos las claves y los nombres legibles desde el diccionario
        clave_origen, nombre_origen = menu_unidades[opcion_origen]
        clave_destino, nombre_destino = menu_unidades[opcion_destino]

        # Realizamos el cálculo usando la función
        calculo = conversor_area(cantidad, clave_origen, clave_destino)

        print(f"\nResultado: {cantidad} {nombre_origen} equivalen a {calculo} {nombre_destino}")

        # Pausa para poder leer el resultado antes de que el menú vuelva a cargar
        input("\nPresiona ENTER para realizar otra conversión...")

    except ValueError:
        print("\nError: Por favor, ingresa un valor numérico válido para la cantidad.")
        input("\nPresiona ENTER para volver a intentarlo...")
```

```python
def conversor_area(valor, clave_origen, clave_destino):
    factores_conversion = {
        "m2": 1.0,              # 1 m² = 1 m²
        "ft2": 0.092903,        # 1 pie² = 0.092903 m²
        "v2": 0.698896,         # 1 vara² = 0.698896 m²
        "yd2": 0.836127,        # 1 yarda² = 0.836127 m²
        "tz": 436.81,           # 1 tarea = 436.81 m²
        "mz": 6988.96,          # 1 manzana = 6988.96 m²
        "ha": 10000.0           # 1 hectárea = 10000 m²
    }

    # Convertir a m²
    valor_en_m2 = valor * factores_conversion[clave_origen]
    # Convertir de m² a la unidad destino
    resultado = valor_en_m2 / factores_conversion[clave_destino]

    return round(resultado, 4)

menu_unidades = {
    "1": ("m2", "Metro Cuadrado"),
    "2": ("ft2", "Pie Cuadrado"),
    "3": ("v2", "Vara Cuadrada"),
    "4": ("yd2", "Yarda Cuadrada"),
    "5": ("tz", "Tareas"),
    "6": ("mz", "Manzanas"),
    "7": ("ha", "Hectárea")
}


def menu():
    while True:
        print("\n--- CONVERSOR DE ÁREA ---")
        for num, (clave, nombre) in menu_unidades.items():
            print(f"{num}. {nombre}")
        print("0. Salir")

        opcion_origen = input("Unidad de origen: ").strip()

        if opcion_origen == "0":
            print("¡Hasta luego!")
            break

        if opcion_origen not in menu_unidades:
            print("Opción no válida, intenta de nuevo.")
            continue

        opcion_destino = input("Unidad de destino: ").strip()

        if opcion_destino not in menu_unidades:
            print("Opción no válida, intenta de nuevo.")
            continue

        try:
            cantidad = float(input("Cantidad a convertir: "))

            clave_origen, nombre_origen = menu_unidades[opcion_origen]
            clave_destino, nombre_destino = menu_unidades[opcion_destino]

            resultado = conversor_area(cantidad, clave_origen, clave_destino)

            print(f"\n{cantidad} {nombre_origen} = {resultado} {nombre_destino}")

        except ValueError:
            print("Error: ingresa un número válido.")


menu()
```

```python
class Estudiante:
    def __init__(self, nombre, edad, tiempo_traslado_min, horas_telefono, materias):
        self.nombre = nombre
        self.edad = edad
        self.tiempo_traslado_min = tiempo_traslado_min
        self.horas_telefono = horas_telefono
        self.materias = materias

    # Obtener notas por materia que tiene el estudiante
    def notas_materia(self, materia):
        # Usando filter para encontrar la materia
        materia_encontrada = list(filter(
            lambda m: m["materia"] == materia,
            self.materias
        ))

        if materia_encontrada:
            # Usando map para extraer las notas
            return list(map(
                lambda clave: materia_encontrada[0][clave],
                ["lab1", "lab2", "parcial"]
            ))
        return None

    def promedio_general(self):
        promedios_materias = list(map(
            lambda m: m["lab1"] * 0.30 + m["lab2"] * 0.30 + m["parcial"] * 0.40,
            self.materias
        ))
        return sum(promedios_materias) / len(promedios_materias)

    def aprobo(self, nota_minima=6.0):
        promedio = self.promedio_general()
        return promedio >= nota_minima

    def materias_aprobadas(self, nota_minima=6.0):
        return list(filter(
            lambda m: (m["lab1"] * 0.30 + m["lab2"] * 0.30 + m["parcial"] * 0.40) >= nota_minima,
            self.materias
        ))


    def promedio_por_materia(self):
        return list(map(
            lambda m: {
                "materia": m["materia"],
                "promedio": m["lab1"] * 0.30 + m["lab2"] * 0.30 + m["parcial"] * 0.40
            },
            self.materias
        ))

    def to_dict(self):
        """Convierte el estudiante a diccionario"""
        return {
            "nombre": self.nombre,
            "edad": self.edad,
            "tiempo_traslado_min": self.tiempo_traslado_min,
            "horas_telefono": self.horas_telefono,
            "materias": self.materias,
            "promedio_general": self.promedio_general(),
            "aprobado": self.aprobo()
        }

class SistemaAcademico:
    def __init__(self):
        self.estudiantes = []
        self.estadistica = Estadistica()

    def cargar_estudiantes(self, datos):
        """Carga estudiantes desde un dataset de diccionarios usando map"""
        self.estudiantes = list(map(
            lambda dato: Estudiante(
                dato["nombre"],
                dato["edad"],
                dato["tiempo_traslado_min"],
                dato["horas_telefono"],
                dato["materias"]
            ),
            datos
        ))

    def reporte_general(self):
        """Reporte general de edad, traslado y telefono usando map y filter"""
        # Usando map para extraer datos
        edades = list(map(lambda e: e.edad, self.estudiantes))
        tiempos_traslado = list(map(lambda e: e.tiempo_traslado_min, self.estudiantes))
        horas_telefono = list(map(lambda e: e.horas_telefono, self.estudiantes))

        print("\n=== REPORTE GENERAL ===")
        print(f"Edad - Promedio: {self.estadistica.media(edades):.2f}, Desviacion: {self.estadistica.desviacion(edades):.2f}")
        print(f"Tiempo traslado - Promedio: {self.estadistica.media(tiempos_traslado):.2f}, Desviacion: {self.estadistica.desviacion(tiempos_traslado):.2f}")
        print(f"Horas telefono - Promedio: {self.estadistica.media(horas_telefono):.2f}, Desviacion: {self.estadistica.desviacion(horas_telefono):.2f}")

    def reporte_campo(self, campo):
        """Reporte de un campo especifico usando map y filter"""
        # Usando map para extraer valores
        valores = list(map(
            lambda e: getattr(e, campo, None),
            self.estudiantes
        ))
        # Filtrar valores None
        valores = list(filter(lambda x: x is not None, valores))

        if valores:
            print(f"\n=== REPORTE DE {campo.upper()} ===")
            print(f"Media: {self.estadistica.media(valores):.2f}")
            print(f"Desviacion: {self.estadistica.desviacion(valores):.2f}")
            print(f"Moda: {self.estadistica.moda(valores)}")
            print(f"Rango: {self.estadistica.rango(valores)}")
            print(f"Frecuencia: {self.estadistica.frecuencia(valores)}")
        else:
            print("Campo no encontrado")

    def reporte_componente(self, componente):
        """Reporte de un componente de notas (lab1, lab2, parcial) usando map y filter"""
        # Usando map y filter para extraer todas las notas de ese componente
        notas = list(filter(
            lambda x: x is not None,
            list(map(
                lambda e: list(map(
                    lambda m: m.get(componente, None),
                    e.materias
                )),
                self.estudiantes
            ))
        ))
        # Aplanar la lista usando reduce
        notas_planas = reduce(lambda x, y: x + y, notas, [])
        # Filtrar None
        notas_limpias = list(filter(lambda x: x is not None, notas_planas))

        if notas_limpias:
            print(f"\n=== REPORTE DE {componente.upper()} ===")
            print(f"Media: {self.estadistica.media(notas_limpias):.2f}")
            print(f"Desviacion: {self.estadistica.desviacion(notas_limpias):.2f}")
            print(f"Moda: {self.estadistica.moda(notas_limpias)}")
            print(f"Rango: {self.estadistica.rango(notas_limpias)}")
        else:
            print("Componente no encontrado")

    def notas_por_materia(self):
        """Promedio de notas por materia usando map, filter y reduce"""
        # Obtener todas las materias únicas
        todas_materias = list(map(
            lambda e: [m["materia"] for m in e.materias],
            self.estudiantes
        ))
        materias_unicas = list(set(reduce(lambda x, y: x + y, todas_materias, [])))

        # Calcular promedios por materia
        promedios_por_materia = {}
        for materia in materias_unicas:
            # Filtrar estudiantes que tienen esta materia
            notas_materia = list(filter(
                lambda x: x is not None,
                list(map(
                    lambda e: e.notas_materia(materia),
                    self.estudiantes
                ))
            ))
            # Calcular promedio de todas las notas de esta materia
            todas_notas = reduce(lambda x, y: x + y, notas_materia, [])
            if todas_notas:
                promedio = sum(todas_notas) / len(todas_notas)
                promedios_por_materia[materia] = promedio

        print("\n=== PROMEDIO DE NOTAS POR MATERIA ===")
        for materia, promedio in promedios_por_materia.items():
            print(f"{materia}: {promedio:.2f}")

    def estudiantes_aprobados(self, nota_minima=6.0):
        """Ver estudiantes aprobados usando filter"""
        return list(map(
            lambda e: (e.nombre, e.promedio_general()),
            filter(lambda e: e.aprobo(nota_minima), self.estudiantes)
        ))

    def estudiantes_reprobados(self, nota_minima=6.0):
        """Ver estudiantes reprobados usando filter"""
        return list(map(
            lambda e: (e.nombre, e.promedio_general()),
            filter(lambda e: not e.aprobo(nota_minima), self.estudiantes)
        ))

    def resumen_por_materia(self):
        """Resumen estadistico completo por materia usando map y reduce"""
        # Obtener todas las materias únicas
        todas_materias = list(map(
            lambda e: [m["materia"] for m in e.materias],
            self.estudiantes
        ))
        materias_unicas = list(set(reduce(lambda x, y: x + y, todas_materias, [])))

        print("\n=== RESUMEN ESTADISTICO POR MATERIA ===")
        for materia in materias_unicas:
            # Obtener todas las notas de esta materia
            notas = list(map(
                lambda e: list(filter(
                    lambda x: x is not None,
                    e.notas_materia(materia) or []
                )),
                self.estudiantes
            ))
            # Aplanar la lista
            notas_planas = reduce(lambda x, y: x + y, notas, [])

            if notas_planas:
                print(f"\n{materia}:")
                print(f"  Media: {self.estadistica.media(notas_planas):.2f}")
                print(f"  Desviacion: {self.estadistica.desviacion(notas_planas):.2f}")
                print(f"  Moda: {self.estadistica.moda(notas_planas)}")
                print(f"  Rango: {self.estadistica.rango(notas_planas)}")


    def mostrar_estudiantes(self):
        """Muestra todos los estudiantes como diccionarios completos"""
        print("\n=== LISTA DE ESTUDIANTES ===")
        print("="*80)

        # Usar map para convertir cada estudiante a diccionario
        estudiantes_dict = list(map(lambda e: e.to_dict(), self.estudiantes))

        # Mostrar cada estudiante como JSON formateado
        for i, estudiante in enumerate(estudiantes_dict, 1):
            print(json.dumps(estudiante, ensure_ascii=False, indent=2))
            print("-"*80)

    def estudiantes_mayores_edad(self, edad_minima=18):
        """Filtra estudiantes mayores de cierta edad usando filter"""
        return list(filter(lambda e: e.edad >= edad_minima, self.estudiantes))

    def promedio_general_grupo(self):
        """Promedio general de todos los estudiantes usando reduce"""
        promedios = list(map(lambda e: (e.nombre, e.promedio_general()), self.estudiantes))

        print("\n=== PROMEDIO GENERAL DEL GRUPO ===")
        for nombre, promedio in promedios:
            print(f"  {nombre}: {promedio:.2f}")

        solo_promedios = list(map(lambda p: p[1], promedios))
        promedio_grupo = reduce(lambda x, y: x + y, solo_promedios, 0) / len(solo_promedios) if solo_promedios else 0

        return promedio_grupo

    def ver_aprobados_reprobados(self, nota_minima=6.0):
        """Muestra estudiantes aprobados y reprobados"""
        aprobados = self.estudiantes_aprobados(nota_minima)
        reprobados = self.estudiantes_reprobados(nota_minima)

        print("\n" + "="*50)
        print("RESUMEN DE APROBADOS Y REPROBADOS")
        print("="*50)

        # Aprobados
        print(f"\nAPROBADOS ({len(aprobados)} estudiantes):")
        if aprobados:
            for nombre, promedio in aprobados:
                print(f"  - {nombre}: {promedio:.2f}")
        else:
            print("  Ningun estudiante aprobo")

        # Reprobados
        print(f"\nREPROBADOS ({len(reprobados)} estudiantes):")
        if reprobados:
            for nombre, promedio in reprobados:
                print(f"  - {nombre}: {promedio:.2f}")
        else:
            print("  Ningun estudiante reprobo")

        # Estadisticas adicionales
        total = len(self.estudiantes)
        if total > 0:
            tasa_aprobacion = (len(aprobados) / total) * 100
            print(f"\nTasa de aprobacion: {tasa_aprobacion:.1f}%")

    def promedios_por_estudiante(self):
        """Muestra el promedio por materia y el promedio general de cada estudiante"""
        print("\n=== PROMEDIO GENERAL POR ESTUDIANTE ===")
        for e in self.estudiantes:
            print(f"\n{e.nombre}")
            print("-" * len(e.nombre))
            for resultado in e.promedio_por_materia():
                print(f"  {resultado['materia']:<35}: {resultado['promedio']:.2f}")
            print(f"  {'Promedio general':<35}: {e.promedio_general():.2f}")

# DATASET: estudiantes
datos_estudiantes = [
    {"nombre": "Jimmy Roberto Martinez Gomez", "edad": 19,
        "tiempo_traslado_min": 30, "horas_telefono": 6,
        "materias": [
            {"materia": "Administracion de Bases de datos II", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion Computacional III", "lab1": 9.0, "lab2": 9.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 8.0, "lab2": 8.3, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 10, "lab2": 10, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 10, "lab2": 10, "parcial": 0},
        ]
    },
    {"nombre": "Wilfredo Josue Ortez Gaitan", "edad": 20,
        "tiempo_traslado_min": 50, "horas_telefono": 6,
        "materias": [
            {"materia": "Administracion de Bases de datos II", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion Computacional III", "lab1": 9, "lab2": 9, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 8, "lab2": 8.3, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 10, "lab2": 10, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 10, "lab2": 10, "parcial": 0},
        ]
    },
    {"nombre": "Miguel Alejandro Gómez Saravia", "edad": 19,
        "tiempo_traslado_min": 15, "horas_telefono": 4,
        "materias": [
            {"materia": "Bases de datos II", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
        ]
    },
    {"nombre": "Josue Stiven Funes Benavides", "edad": 19,
        "tiempo_traslado_min": 30, "horas_telefono": 5,
        "materias": [
            {"materia": "Bases de datos II", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
        ]
    },
    {"nombre": "Jefferson Steven Castillo Leiva", "edad": 19,
        "tiempo_traslado_min": 15, "horas_telefono": 2.3,  # 2h 18min
        "materias": [
            {"materia": "Bases de datos II", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
        ]
    },
    {"nombre": "Ramael Ernesto Cordova Colindres", "edad": 22,
        "tiempo_traslado_min": 7, "horas_telefono": 13.52,  # 13h 31min
        "materias": [
            {"materia": "Bases de datos II", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
        ]
    },
    {"nombre": "Carlos Daniel Parada Avalos", "edad": 25,
        "tiempo_traslado_min": 10, "horas_telefono": 5.52,  # interpretado como 5h 31min
        "materias": [
            {"materia": "Bases de datos II", "lab1": 8.0, "lab2": 9.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 8.0, "lab2": 9.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 8.0, "lab2": 9.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 8.0, "lab2": 9.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 8.0, "lab2": 9.0, "parcial": 0},
        ]
    },
    {"nombre": "Yeimy Julissa Parada Chavez", "edad": 20,
        "tiempo_traslado_min": 45, "horas_telefono": 3,
        "materias": [
            {"materia": "Bases de datos II", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 9.0, "lab2": 10.0, "parcial": 0},
        ]
    },
    {"nombre": "Madison Odalis Valle Juarez", "edad": 19,  # no se proporciono
        "tiempo_traslado_min": 30, "horas_telefono": 9,
        "materias": [  # asumido mismo set de materias del grupo; solo dijo "todas son 10"
            {"materia": "Bases de datos II", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
        ]
    },
    {"nombre": "Jennifer Maricela Renderos Guevara", "edad": 21,
        "tiempo_traslado_min": 60, "horas_telefono": 3,
        "materias": [
            {"materia": "Bases de datos II", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Programacion III", "lab1": 8.0, "lab2": 8.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Estadistica Computacional", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
        ]
    },
    {"nombre": "Kimberly Michelle Castro Zelaya", "edad": 18,
        "tiempo_traslado_min": 20, "horas_telefono": 9,
        "materias": [  # solo se proporciono esta materia
            {"materia": "Programacion Computacional III", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
        ]
    },
    {"nombre": "Samuel Alexander Portillo Ceron", "edad": 20,
        "tiempo_traslado_min": 80,  # 1:20 interpretado como 1h20min
        "horas_telefono": 6,
        "materias": [
            {"materia": "Estadistica", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Diseno de Circuitos Digitales", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Ingles Basico", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
            {"materia": "Base de Datos II", "lab1": 10.0, "lab2": 10.0, "parcial": 0},
        ]
    },
]

# MENU PRINCIPAL
def menu():
    sistema = SistemaAcademico()
    sistema.cargar_estudiantes(datos_estudiantes)

    while True:
        print("\n========== SISTEMA ESTADISTICO ACADEMICO ==========")
        print("1. Reporte general (edad, traslado, telefono)")
        print("2. Reporte de un campo especifico")
        print("3. Promedio de notas por materia")
        print("4. Ver aprobados y reprobados")
        print("5. Ver todos los estudiantes")
        print("6. Resumen estadistico completo por materia")
        print("7. Promedio general por estudiante")
        print("8. Promedio general del grupo")
        print("0. Salir")

        opcion = input("Elige una opcion: ")

        if opcion == "1":
            sistema.reporte_general()

        elif opcion == "2":
            print("Campos disponibles:\n1-edad\n2-tiempo traslado\n3-horas telefono\n4-lab1\n5-lab2\n6-parcial")
            campo = input("Escribe el numero del campo: ")

            if campo == "1":
                sistema.reporte_campo("edad")
            elif campo == "2":
                sistema.reporte_campo("tiempo_traslado_min")
            elif campo == "3":
                sistema.reporte_campo("horas_telefono")
            elif campo == "4":
                sistema.reporte_componente("lab1")
            elif campo == "5":
                sistema.reporte_componente("lab2")
            elif campo == "6":
                sistema.reporte_componente("parcial")
            else:
                print("Campo no valido.")

        elif opcion == "3":
            sistema.notas_por_materia()

        elif opcion == "4":
            sistema.ver_aprobados_reprobados()

        elif opcion == "5":
            sistema.mostrar_estudiantes()

        elif opcion == "6":
            sistema.resumen_por_materia()

        elif opcion == "7":
            sistema.promedios_por_estudiante()

        elif opcion == "8":
            promedio_grupo = sistema.promedio_general_grupo()
            print(f"\nPromedio general del grupo: {promedio_grupo:.2f}")


        elif opcion == "0":
            print("Saliendo del sistema...")
            break

        else:
            print("Opcion invalida, intenta de nuevo.")

# Correr el sistema
menu()
```
